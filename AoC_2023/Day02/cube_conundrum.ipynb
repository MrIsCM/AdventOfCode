```python
import re
```

# Read data

```python
with open('input.txt') as f:
    data = f.read().split('\n')
```

## Parse data

```python
# Split lines by :, , or ;
pattern = re.compile(r'[:,;]')
data_processes = []
for line in data:
    data_processes.append(re.split(pattern, line))
```

```python
data_processes
```

```python
parsed_data = []
for game in data_processes:
    game_list = []
    for item in game:
        game_list.append(item.strip())
    
    parsed_data.append(game_list)
```

# Part 1

```python
# 12 red cubes, 13 green cubes, and 14 blue
red_max = 12
green_max = 13
blue_max = 14

max_per_color = {
    'red' : 12,
    'green' : 13,
    'blue' : 14
}
```

```python
sum = int(len(parsed_data)*(len(parsed_data)+1)/2)
for idx, game in enumerate(parsed_data):
    for item in game[1:]:
        num, color = item.split(' ')
        if int(num) > max_per_color[color]:
            sum -= idx + 1
            break

sum
```

# Part 2

```python
suma2 = 0
for idx, game in enumerate(parsed_data):
    maxes = {
        'red': 0,
        'blue': 0,
        'green':0
    }
    # print(f"Game #{idx+1}:")
    for item in game[1:]:
        # print(item)
        num, color = item.split(' ')
        if maxes[color] < int(num):
            maxes[color] = int(num)

    suma2 += maxes['red']*maxes['blue']*maxes['green']

suma2
```
