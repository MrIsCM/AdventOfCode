# Part 1

```python
with open("input.txt") as f:
    data = f.read().split('\n')
```

```python
digits = list(map(str, range(10)))
digits
```

```python
sum = 0
for line in data:
    for char in line:
        if char in digits:
            sum += 10*int(char)
            break

    for char in line[::-1]:
        if char in digits:
            sum += int(char)
            break
sum
```

# Part 2

```python
def find_nums(patterns, line, inverse=False, verbose=False):
    '''
    Finds the first ocurrence of a pattern in data
    '''
    
    temp = ''
    if not inverse:
        for idx, char in enumerate(line):
            temp += char
            if verbose:
                print(f"    Temp : {temp}")
            for pattern in patterns:
                if pattern in temp:
                    return idx, pattern
    
    if inverse:
        for idx, char in enumerate(line[::-1]):
            temp += char
            temp_inv = temp[::-1]
            if verbose:
                print(f"    Inv  : {temp_inv}")
            for pattern in patterns:
                if pattern in temp_inv:
                    return len(line)-1-idx, pattern
```

```python
digits_written = ['zero', 'one', 'two', 'three', 'four', 'five', 'six', 'seven', 'eight', 'nine']
```

```python
patterns = digits.copy()
patterns.extend(digits_written)
print(patterns)
```

```python
num_dict = {w : int(d) for w, d in zip(digits_written, digits)}

for digit in digits:
    num_dict[digit] = int(digit)
```

```python
sum2 = 0
for line in data:
    # print(line)
    l_idx, l_val = find_nums(patterns, line, inverse=False, verbose=False) # type: ignore
    r_idx, r_val = find_nums(patterns, line, inverse=True, verbose=False) # type: ignore

    sum2 += num_dict[l_val]*10 + num_dict[r_val]
```
