# Scratchcards

```python
import re
```

# Read and Parse Data

```python
# Read data
with open('input.txt') as f:
    data = f.read().split('\n')
```

```python
# Parse
pattern = re.compile(r':\s*|\|\s*')
'''
Explanation (short):
- Compiles a regular expression that matches either:
    - ":" followed by optional whitespace (\s*), or
    - a literal "|" followed by optional whitespace.
- r'...' is a raw string so backslashes are not interpreted by Python.
- The alternation operator | inside the regex is used between the two alternatives; the pipe character to match literally is escaped as \|.
- Using re.compile speeds repeated use (e.g., re.split(pattern, line)).
Example:
- re.split(pattern, "Card 209: 61 7 ... | 85 4 ...") -> ["Card 209", "61 7 ...", "85 4 ..."]
'''

parsed_data = []
for line in data:
    if not line.strip():
        continue
    parts = [p.strip() for p in re.split(pattern, line) if p.strip()]
    parsed_data.append(parts)
```

```python
# Finish parsing using a dict (bc I like dicts)
data_dict = {}
for game_id, winning_nums, my_nums in parsed_data:
    data_dict[game_id] = {
        'winning nums': set(map(int, re.split(r'\s+', winning_nums.strip()))),
        'my nums' : set(map(int, re.split(r'\s+', my_nums.strip())))
        }
```

# Part 1

```python
total_points = 0
for _, game_data in data_dict.items():
    # Find the intersection between the winning nums and my nums
    results = game_data['winning nums'].intersection(game_data['my nums'])
    l = len(results)

    if l == 0:
        continue

    else:
        total_points += 2**(l-1)

total_points
```

# Part 2

```python
# Dict proved to be usefull for part 2
data_dict_2 = {}
for game_id, winning_nums, my_nums in parsed_data:
    _, id_num = re.split(r'\s+', game_id)
    data_dict_2[int(id_num)] = {
        'n of cards' : 1,
        'winning nums': set(map(int, re.split(r'\s+', winning_nums.strip()))),
        'my nums' : set(map(int, re.split(r'\s+', my_nums.strip())))
        }
```

```python
total_n_cards = 0
for card_id, card_info in data_dict_2.items():
    n = card_info['n of cards']
    print(f"Processing card {card_id}: I have {n} copies")
    print("  winning_nums =", card_info['winning nums'])
    print("  my_nums      =", card_info['my nums'])
    inter = card_info['winning nums'].intersection(card_info['my nums'])
    print("  matches =", len(inter))
    if inter:
        for j in range(1, len(inter) + 1):
            if card_id+j >= len(data_dict_2):
                continue
            target = card_id + j
            target_n = data_dict_2[card_id+j]['n of cards']
            print(f"    will add {n} to card {target} (current n_of_cards={target_n})")
    total_n_cards += card_info['n of cards']
    winning_nums = card_info['winning nums']
    my_nums = card_info['my nums']

    matches = len(winning_nums.intersection(my_nums))
    for i in range(matches):
        if card_id+1+i > len(data_dict_2):
            continue
        data_dict_2[card_id+i+1]['n of cards'] += data_dict_2[card_id]['n of cards']

total_n_cards
```
