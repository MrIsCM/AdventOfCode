# Read and Parse Data

```python
def read_parse(path):
    with open(path) as f:
        data = f.read().splitlines()

    parsed_data = {}
    for row in data:
        name, values = row.split(':', 1)
        parsed_data[name.strip()] = [int(v) for v in values.split()]

    return parsed_data['Time'], parsed_data['Distance']
```

```python
data = read_parse('input.txt')
test = read_parse('test.txt') 
```

# Part 1

We have to solve for $t_s$ (Time stopped), considering:

$$
t_S + t_m = t_T \\
d = t_m \cdot v \\
$$

and

$$ 
v = t_s \\
d > d_o
$$

where: 
- $t_S$ is the time stopped
- $t_m$ is the time moving
- $t_T$ is the total time
- $d$ is the distance traveled
- $v$ is the speed
- $d_o$ is the objetive distance

Reordering:

$$
t_s = t_T - t_m \\
t_m = t_T - t_s \\
$$

And subs:
$$
d = (t_T - t_s)\cdot t_s \Rightarrow
d = t_T\cdot t_s - t_s^2 
$$

Then:
$$
t_T\cdot t_s - t_s^2 > d_o \Rightarrow \\
-t_s^2 + t_T\cdot t_s - d_0 > 0
$$

Solving the equality:

$$
t_s = \frac{-t_T \pm \sqrt{t_T^2 - 4 \cdot d_0}}{-2}
$$

```python
def solve_2nd_deg_eq(a, b, c):
    return (-b + math.sqrt(b**2 - 4*a*c))/(2*a), (-b - math.sqrt(b**2 - 4*a*c))/(2*a)
```

```python
def solve(data):
    return [solve_2nd_deg_eq(-1, total_time, -distance) for total_time, distance in zip(data[0], data[1])]
```

```python
def num_ways(solutions):
    prod = 1
    for sol in solutions:
        left = int(sol[0])
        right = int(sol[1]) if int(sol[1]) != sol[1] else int(sol[1]) -1  
        prod *= (right-left)

    return prod
```
