# Read & Process Data

```python
with open("test.txt") as f:
    test_data = f.readlines()
```

```python
with open("input.txt") as f:
    data = f.readlines()
```

```python
def parse(data):
    parsed = []
    for line in data:
        parsed.append(list(map(int, line.strip())))

    return parsed
```

```python
test_data = parse(test_data)
data = parse(data)
```

# Part 1

```python
def biggest_2_dig(data, verbose=False):
    
    suma = 0
    for line in data:
        if verbose:
            print(f"Processing line: {line}")
        first = max(line)
        first_pos = line.index(first)
        if verbose:
            print(f"  First: {first} at position {first_pos}")

        if first_pos == len(line) - 1:
            second = max(line[:-1])
            second_pos = line[:-1].index(second)
            if verbose:
                print(f"  Second (before first): {second} at position {second_pos}")

            number = second*10 + first

        else:
            second = max(line[first_pos+1:])
            second_pos = line[first_pos+1:].index(second) + first_pos + 1
            if verbose:
                print(f"  Second (after first): {second} at position {second_pos}")

            number = first*10 + second

        if verbose:
            print(f"  Number formed: {number}")
        suma += number
        if verbose:
            print(f"  Running sum: {suma}\n")

    return suma
```

```python
biggest_2_dig(test_data)
```

```python
biggest_2_dig(data)
```

# Part 2

```python
n = 12
suma = 0
verbose = True

for line in test_data:
    number = ''
    line_copy = line.copy()
    if verbose:
        print(f" Line: {line}")
    pos = 0
    prev_pos = 0
    for i in range(n-1):
        # Find max digit allowing space for the rest
        digit = max(line_copy[:-(n-i)])
        pos = line_copy.index(digit)
        
        number += str(digit)
        if verbose:
            print(f"    Current max digit: {digit} in position: {prev_pos + pos}")
            print(f"    Current number: {number}")
        
        line_copy = line_copy[pos + 1:]
        prev_pos += pos + 1
    
    last_num = max(line_copy[prev_pos:])
    number += str(last_num)
    if verbose:
            print(f"    Current max digit: {last_num} in position: {prev_pos + 1 + pos}")
            print(f"    Current number: {number}")
    
    suma += int(number)

suma
```
